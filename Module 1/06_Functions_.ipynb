## Introduction

* Suppose we need to calculate profit for many products
* We could write the formula separately for each product, but that would be repetitive and difficult to maintain
* If the formula changes later, we would need to change it everywhere
* A ***function*** solves this problem by keeping the logic in one place
* We write it once and use it whenever required

`````python
# Without a function: the formula is repeated for each product
profit1 = 60000 - 42000
profit2 = 45000 - 30000
profit3 = 80000 - 52000

print(profit1, profit2, profit3)
`````

## What is a Function?

* A ***function*** is a reusable block of code that performs a specific task
* The structure of a function is as follows:

````
def function_name():
    code_to_run
````

* We define a function using the keyword `def`
* After defining the function, we can call it by writing the function name followed by parentheses

## Defining and Calling a Function

Example: We define a function below to print a welcome message

`````python
def welcome_message():
    print("Welcome to Python for Data Analytics")
`````

After defining a function, we have to ***call*** the function to use it. We can call the function as follows:

`````python
welcome_message()
`````

**Note**: Defining a function does not run it. Nothing is printed until the function is called. A function can be called as many times as needed.

`````python
welcome_message()
welcome_message()
`````

## Function with Input

* A function can take inputs
* These inputs are called ***parameters***

Example:

`````python
def add_numbers(a, b):
    print(a + b)
`````

Here:

* `a` and `b` are the ***parameters***
* The function prints the sum of the two input parameters

**Note**: The slide names this function `sum`, but `sum` is already a built-in Python function, so we use `add_numbers` instead.

Calling the function:

`````python
add_numbers(10, 20)
add_numbers(500, 250)
`````

## Function with Return Value

* In the previous examples, we were printing the result of functions
* Instead of printing, we can ***return*** the result using `return`
* Returning is useful because the value can be stored and used later for other calculations

Example:

`````python
def calculate_revenue(units_sold, selling_price):
    revenue = units_sold * selling_price
    return revenue
`````

Calling the function:

`````python
product_revenue = calculate_revenue(100, 500)
print(product_revenue)
`````

## Print vs Return

print_vs_return.png

Example: The value returned by a function can be used in further calculations.

`````python
def revenue_with_print(units_sold, selling_price):
    print(units_sold * selling_price)

def revenue_with_return(units_sold, selling_price):
    return units_sold * selling_price

a = revenue_with_print(100, 500)
b = revenue_with_return(100, 500)

print("a =", a)          # None: print() does not give back a value
print("b =", b)          # 50000
print("b - 42000 =", b - 42000)   # the returned value can be reused
`````

## Example

Write a function that takes revenue and total cost as input and returns the profit.

`````python
def calculate_profit(revenue, total_cost):
    profit = revenue - total_cost
    return profit
`````

`````python
profit = calculate_profit(60000, 42000)
print("Profit:", profit)
`````

The same function can be reused for many products:

`````python
print(calculate_profit(45000, 30000))
print(calculate_profit(80000, 52000))
print(calculate_profit(25000, 27000))
`````

## Function for Business Decision Rule

* Functions can also be used for decision rules
* Example: The following function checks whether the sales target was achieved or not and returns a status
* This can be used for employees, branches, products or regions

`````python
def check_target_status(sales, target):
    if sales >= target:
        return "Target Achieved"
    else:
        return "Target Not Achieved"
`````

`````python
print(check_target_status(75000, 60000))
print(check_target_status(45000, 60000))
`````

## Guidelines for Function Names

Good function names should be:

1. ***Meaningful***
2. ***Short and clear***
3. ***Written in lowercase***
4. ***Words separated using underscores***

Good naming examples:

* `calculate_revenue()`
* `calculate_profit()`
* `check_target_status()`
* `apply_discount()`

Poor naming examples:

* `abc()`
* `func1()`
* `do_this()`
* `x()`

## More Business Examples (Extra)

**Note**: This section is not in the lecture slides.

Example 1: Apply a discount (this was mentioned in the agenda)

The function takes a price and a discount percentage and returns the discounted price.

`````python
def apply_discount(price, discount_percent):
    discount = price * discount_percent / 100
    return price - discount

print(apply_discount(1000, 10))
print(apply_discount(25000, 15))
`````

Example 2: Use one function inside another

`````python
def calculate_revenue(units_sold, selling_price):
    return units_sold * selling_price

def calculate_profit(revenue, total_cost):
    return revenue - total_cost

revenue = calculate_revenue(120, 500)
profit = calculate_profit(revenue, 42000)

print("Revenue:", revenue)
print("Profit:", profit)
`````

Example 3: Use a function with a loop to calculate profit for many products

`````python
revenues = [60000, 45000, 80000]
costs = [42000, 30000, 52000]

for i in range(len(revenues)):
    profit = calculate_profit(revenues[i], costs[i])
    print("Product", i + 1, "profit:", profit)
`````

Example 4: A function with a default parameter value

If a value is not given when calling the function, the default value is used.

`````python
def apply_gst(price, gst_percent=18):
    return price + price * gst_percent / 100

print(apply_gst(1000))       # uses default 18%
print(apply_gst(1000, 5))    # uses 5%
`````

Example 5: Classify a customer using a function

`````python
def customer_category(spending):
    if spending >= 100000:
        return "Premium"
    elif spending >= 50000:
        return "Regular"
    else:
        return "Low-value"

for amount in [120000, 65000, 30000]:
    print(amount, "->", customer_category(amount))
`````

## Practice Questions


1. Write a function `calculate_margin(revenue, cost)` that returns profit margin as a percentage of revenue.
2. Write a function `is_high_rating(rating)` that returns `True` if the rating is 4.0 or more, otherwise `False`.
3. Write a function `average(values)` that takes a list and returns its average.

`````python
# Answers
def calculate_margin(revenue, cost):
    return (revenue - cost) / revenue * 100

def is_high_rating(rating):
    return rating >= 4.0

def average(values):
    return sum(values) / len(values)

print(calculate_margin(60000, 42000))
print(is_high_rating(4.3))
print(average([4.5, 4.2, 3.9]))
`````
