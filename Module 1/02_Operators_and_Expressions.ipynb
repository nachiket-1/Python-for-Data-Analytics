## What is an Operator?

An ***operator*** is a symbol that performs an operation on values.

Example:

```python
units_sold = 120
selling_price = 500
revenue = units_sold * selling_price

print(revenue)
```

## Here:

* `*` is the ***multiplication operator***
* `units_sold` and `selling_price` are the ***variables***
* `revenue` stores the result

## What is an Expression?

An ***expression*** is a combination of values, variables, and operators that gives a result.

Example:

```python
revenue = 60000
total_cost = 45000

profit = revenue - total_cost
print(profit)
```

## Here:

* `revenue - total_cost` is an ***expression***
* Python evaluates the expression
* The result is stored in `profit`

## Arithmetic Operators

Arithmetic operators are used for mathematical calculations.



arithmetic_operators.png



```python
sales = 50000
tax = 9000
revenue = 80000
cost = 55000
units = 40
price = 75
profit = 25000

print(sales + tax)       # Addition
print(revenue - cost)    # Subtraction
print(units * price)     # Multiplication
print(profit / revenue)  # Division
print(105 // 10)         # Floor division
print(105 % 10)          # Remainder
print(2 ** 3)            # Power
```

**Note**: `/` always gives a float (decimal) result, while `//` gives only the whole-number part (rounded down).

```python
print(10 / 2)     # 5.0
print(10 // 2)    # 5
print(-7 // 2)    # -4  (rounds down)
```

## Comparison Operators

* Comparison operators compare two values and return `True` or `False`
* Comparison operators are used when we want to check conditions

comparison_operators.png

```python
sales = 52000
target = 50000
cost = 40000
budget = 45000
rating = 4.2
delay = 3
region = "South"
status = "Active"

print(sales > target)
print(cost < budget)
print(rating >= 4.0)
print(delay <= 2)
print(region == "South")
print(status != "Inactive")
```

## Logical Operators

Logical operators combine multiple conditions.

logical_operators.png

***Boolean vs Logical Operators:***

Logical operators are commonly used on Boolean values (`True` / `False`).

For example: if condition1 `and` condition2 are `True`, perform some action.

```python
sales = 52000
target = 50000
rating = 3.8
customer_active = True

print((sales > target) and (rating >= 4))
print((sales > target) or (rating >= 4))
print(not customer_active)
```

## Assignment Operators

Assignment operators are used to assign or update values.

assignment_operators.png

```python
sales = 50000
sales += 10000
print(sales)      # 60000

sales -= 10000
print(sales)      # 50000

price = 100
price *= 1.10
print(price)      # about 110

budget = 80000
budget /= 2
print(budget)     # 40000.0
```

## Order of Operations

Python follows the mathematical order of operations.

***Order:***

1. Parentheses `()`
2. Power `**`
3. Multiplication, division, remainder
4. Addition and subtraction

Example 1:

```python
result = 10 + 5 * 2
print(result)
```

Example 2:**bold text**

```python
result = (10 + 5) * 2
print(result)
```

## Examples

Example 1: Sales report

```python
units_sold = 150
selling_price = 200
total_cost = 22000

revenue = units_sold * selling_price
profit = revenue - total_cost
profit_margin = profit / revenue

print("Revenue:", revenue)
print("Profit:", profit)
print("Profit margin:", round(profit_margin * 100, 2), "%")
```

Example 2: Bonus eligibility (comparison + logical)

```python
sales = 55000
target = 50000
rating = 4.5

eligible = (sales > target) and (rating >= 4)
print("Eligible for bonus:", eligible)
```

Example 3: Floor division and remainder

```python
total_minutes = 275
hours = total_minutes // 60
minutes = total_minutes % 60

print(hours, "hours", minutes, "minutes")
```

## Basic Business Inputs

```python
# Basic product sales information

product_name = "Wireless Mouse"
units_sold = 120
selling_price = 500
cost_price = 350

print("Product Name:", product_name)
print("Units Sold:", units_sold)
print("Selling Price:", selling_price)
print("Cost Price:", cost_price)
```

## Revenue Calculation

```python
# Revenue calculation

revenue = units_sold * selling_price

print("Revenue:", revenue)
```

## Total Cost and Profit Calculation


```python
# Total cost and profit calculation

total_cost = units_sold * cost_price
profit = revenue - total_cost

print("Total Cost:", total_cost)
print("Profit:", profit)
```

## Profit Margin Calculation

```python
# Profit margin calculation

profit_margin = (profit / revenue) * 100

print("Profit Margin (%):", profit_margin)
```

## Discount Calculation

```python
# Discount calculation

discount_percent = 10

discount_amount = revenue * discount_percent / 100
final_revenue = revenue - discount_amount

print("Original Revenue:", revenue)
print("Discount Amount:", discount_amount)
print("Final Revenue after Discount:", final_revenue)
```

## Profit after Discount

```python
# Profit after discount

profit_after_discount = final_revenue - total_cost

print("Profit after Discount:", profit_after_discount)
```

## Target Achievement Check

```python
# Comparison operator example

sales_target = 50000

target_achieved = final_revenue >= sales_target
#Python compares final_revenue with sales target.
#The output is a Boolean (True or False) value which gets stored in target_achieved

print("Sales Target:", sales_target)
print("Final Revenue:", final_revenue)
print("Target Achieved:", target_achieved)
```

## Logical Operator Example

```python
# Logical operator example
# Product is high-performing if:
# 1. Target is achieved
# 2. Profit margin is at least 20%

high_performing = target_achieved and profit_margin >= 20

print("Target Achieved:", target_achieved)
print("Profit Margin:", profit_margin)
print("High Performing Product:", high_performing)
```

## What Happens If Discount Is Increased?

```python
# Scenario analysis: increase discount to 25%

discount_percent = 25

discount_amount = revenue * discount_percent / 100
final_revenue = revenue - discount_amount
profit_after_discount = final_revenue - total_cost
target_achieved = final_revenue >= sales_target
high_performing = target_achieved and profit_after_discount > 0

print("Discount Percent:", discount_percent)
print("Final Revenue:", final_revenue)
print("Profit after Discount:", profit_after_discount)
print("Target Achieved:", target_achieved)
print("High Performing Product:", high_performing)
```

## Practice Exercise

A product has the following details:

- Product name: Office Chair
- Units sold: 80
- Selling price per unit: ₹2500
- Cost price per unit: ₹1800
- Discount: 12%
- Sales target: ₹160000

Tasks:
1. Calculate revenue before discount
2. Calculate total cost
3. Calculate discount amount
4. Calculate final revenue after discount
5. Calculate profit after discount
6. Check whether the sales target is achieved
7. Check whether the product is profitable

```python
product_name = "Office chair"
units_sold = 80
selling_price_per_unit = 2500
cost_price_per_unit = 1800
dicount_percent = 12
sales_target = 160000

revenue_before_discount = units_sold * selling_price_per_unit
total_cost = units_sold * cost_price_per_unit
discount_amount = revenue_before_discount * dicount_percent / 100
final_revenue_after_discount = revenue_before_discount - discount_amount
profit_after_discount = final_revenue_after_discount - total_cost
target_achieved = final_revenue_after_discount >= sales_target
product_profitable = profit_after_discount > 0

print("Product name:", product_name)
print("Revenue before discount :", revenue_before_discount)
print("Total cost:", total_cost)
print("Discount amount:", discount_amount)
print("Final revenue after discount:", final_revenue_after_discount)
print("Profit after discount:", profit_after_discount)
print("Target achieved:", target_achieved)
print("Product profitable:", product_profitable)
```
