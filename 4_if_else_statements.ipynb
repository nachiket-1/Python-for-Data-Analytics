## Introduction

Business decisions often depend on ***conditions***.

Examples:

* If sales are greater than target, performance is good
* If profit is negative, product needs review
* If stock is low, reorder inventory

If-else statements help Python choose what action to take based on a condition.

## Basic If Statement

* An ***if*** statement runs code only when a condition is true
* After specifying the condition, we have to use a colon `:`
* The statement is then provided on the next line (indented)

Syntax:

```
if condition:
    statement
```

Example:

````python
sales = 75000
target = 60000

if sales >= target:
    print("Target achieved")
````

**Note**: If the condition is `False`, an `if` statement alone does nothing. Nothing is printed below because `sales` is less than `target`.

````python
sales = 45000
target = 60000

if sales >= target:
    print("Target achieved")
````

## If-Else Statement

An ***if-else*** statement gives two possible outcomes.

Syntax:

```
if condition:
    statement_if_true
else:
    statement_if_false
```

Example:

````python
sales = 45000
target = 60000

if sales >= target:
    print("Target achieved")
else:
    print("Target not achieved")
````

## Sales Example

* If monthly sales are greater than or equal to target, performance is satisfactory
* Otherwise, performance needs improvement

````python
monthly_sales = 85000
sales_target = 75000

if monthly_sales >= sales_target:
    print("Performance is satisfactory")
else:
    print("Performance needs improvement")
````

## If-Elif-Else Statement

* When there are more than two outcomes, use ***if-elif-else***
* For example, customer spending may not be simply high or low
* We may want to classify customers as premium, regular, or low-value. In such cases, we use `elif`

Syntax:

```
if condition_1:
    statement_1
elif condition_2:
    statement_2
else:
    statement_3
```

## Customer Categories

Example: Categorizing customers based on annual spending

| Annual Spending | Customer Category |
|:---:|:---:|
| ₹1,00,000 or more | Premium |
| ₹50,000 to ₹99,999 | Regular |
| Below ₹50,000 | Low-value |

````python
spending = 75000

if spending >= 100000:
    category = "Premium"
elif spending >= 50000:
    category = "Regular"
else:
    category = "Low-value"

print(category)
````

**Note**: Python checks the conditions from top to bottom and runs only the first block whose condition is `True`. Try changing `spending` to `120000` and `30000` and run the cell again.

## Logical Operators in Conditions

Multiple conditions can be combined using ***logical operators***.

logical_operators_in_conditions.png

Example (`or`):

A customer receives a discount if:

* Purchase amount is greater than ₹10,000

  ***or***

* Loyalty points are greater than 500

````python
purchase_amount = 8000
loyalty_points = 650

if purchase_amount > 10000 or loyalty_points > 500:
    print("Discount eligible")
else:
    print("Not eligible for discount")
````

Example (`and`): Identify high-performing products

A product is high-performing if:

* Sales greater than 80,000
* Customer rating is more than 4

````python
sales = 95000
sales_target = 80000
customer_rating = 4.4

if sales >= sales_target and customer_rating >= 4.0:
    product_status = "High-performing product"
else:
    product_status = "Needs review"

print(product_status)
````

Example (`not`):

* We can use the `not` operator to check conditions that use Boolean values
* In the following example, we use the `not` operator to check whether a name is not present in the `names` list

````python
names = ['Ravi', 'Suresh', 'Asha']

if 'Aakash' not in names:
    print('Name is missing')
else:
    print('Name is present')
````

##Some More Examples:

## Checking a login status

If-else statements are often used to check whether something is allowed or not.

````python
is_logged_in = True
if is_logged_in:
    print("Access granted")
else:
    print("Please login")
````

Here, `is_logged_in` already stores `True`. So Python directly runs the `if` block.

## Using comparison operators

Comparison operators help Python compare values.

````python
sales = 700
if sales > 500:
    print("High sale")
else:
    print("Normal sale")
````

`>` means greater than. Since 700 is greater than 500, Python prints `High sale`.

## Using `elif` for multiple conditions

`elif` means "else if". It is used when we have more than two possible decisions.

````python
marks = 82
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")
````

Python checks conditions from top to bottom. Since 82 is not greater than

or equal to 90 but is greater than or equal to 75, it prints `Grade B`.

##  Business example: discount eligibility

Suppose a customer gets a discount only if the bill amount is at least 1000.

````python
bill = 1200
if bill >= 1000:
    print("Discount applied")
else:
    print("No discount")
````

The bill is 1200, so the customer is eligible for the discount.

## Calculating discounted price

We can also use if-else to calculate a final value.

````python
price = 1500
if price >= 1000:
    price = price * 0.90
print("Final price:", price)
````

Since the price is at least 1000, Python applies a 10% discount by multiplying the price by 0.90.

##  Using `and`

`and` is used when both conditions must be true.

````python
age = 22
has_id = True
if age >= 18 and has_id:
    print("Entry allowed")
````

Python allows entry only when age is at least 18 and ID is available.

## Using `or`

`or` is used when at least one condition must be true.

````python
payment = "UPI"
if payment == "Cash" or payment == "UPI":
    print("Payment accepted")
````

`==` checks equality. Here, payment is `"UPI"`, so the condition is true and Python prints `Payment accepted`.

## Quick recap

- `if` checks a condition.
- `else` runs when the `if` condition is false.
- `elif` is used when there are multiple conditions.
- `and` means both conditions must be true.
- `or` means at least one condition must be true.
- Indentation is very important in if-else statements.
