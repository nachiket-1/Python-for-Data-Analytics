## What is a List?

A ***list*** is a collection of multiple values stored in one variable.

Example:

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
print(products)
```

## Here:

* `products` is the ***list name***
* The list contains four product names
* Values are placed inside ***square brackets*** `[ ]`
* Each value is separated by a ***comma***

## Why Do We Need Lists?

***Lists are useful for:***

* Product names
* Sales values
* Customer ratings
* Regional revenue
* Employee performance scores
* Monthly demand values

## Why Do We Need Lists? (Continued)

* Lists help us store and manage multiple values
* Without lists, we would need to create many separate variables
* With lists, we can store all related values together

***Without lists:***

```python
sales1 = 50000
sales2 = 65000
sales3 = 48000
sales4 = 72000
```

***With lists:***

```python
sales = [50000, 65000, 48000, 72000]
print(sales)
```

## Creating Lists

Lists can store different types of values.

Thus, a list can store:

1. ***Text values***
2. ***Numbers***
3. ***Decimal values***
4. ***Boolean values***

Examples of Lists:

```python
products = ["Laptop", "Mouse", "Keyboard"]
sales = [50000, 25000, 18000]
ratings = [4.5, 4.2, 3.9]
availability = [True, True, False]

print(products)
print(sales)
print(ratings)
print(availability)
```

**Note**: We can check the data type of a list using `type()`.

```python
print(type(products))
```

## Accessing Values from a List

* List values are accessed using ***index positions***
* Python indexing starts from ***zero***

list_index_table.png

Example:

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
```

To access the first element in the list:

```python
print(products[0])
```

This will access the value at index `0` (Laptop) and print it in the output.

```python
print(products[1])    # Mouse
print(products[2])    # Keyboard
print(products[3])    # Monitor
```

**Note**: Python also allows ***negative indexing***. Index `-1` gives the last element of the list.

```python
print(products[-1])   # Monitor
```

## Updating List Values

List values can be changed using ***index positions***.

Example:

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
print("Before:", products)

products[1] = "Wireless Mouse"
print("After:", products)
```

## Useful List Operations

Common list operations:

useful_list_operations.png

**Note**: The slide uses `list` as a variable name, but `list` is a built-in Python name, so it is better to avoid it. Here we use `my_list` instead.

***1. len()***

Counts the number of values in the list.

```python
print(len([1, 2, 3]))
```

***2. sum()***

Sums up the numeric values in the list.

```python
print(sum([1, 2, 3]))
```

***3. max() and min()***

Find the maximum and minimum values in the list.

```python
print(max([1, 2, 3]))
print(min([1, 2, 3]))
```

***4. append()***

Adds the new value at the end of the list.

```python
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)
```

***5. remove()***

Removes the specified value from the list.

```python
my_list.remove(3)
print(my_list)
```

## Business Examples

Example 1: Sales analysis

```python
sales = [50000, 65000, 48000, 72000]

print("Number of sales values:", len(sales))
print("Total sales:", sum(sales))
print("Highest sales:", max(sales))
print("Lowest sales:", min(sales))
print("Average sales:", sum(sales) / len(sales))
```

Example 2: Customer ratings

ratings = [4.5, 4.2, 3.9]
ratings.append(4.8)

print(ratings)
print("Average rating:", round(sum(ratings) / len(ratings), 2))

Example 3: Managing products

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]

products.append("Printer")
products[1] = "Wireless Mouse"
products.remove("Keyboard")

print(products)
print("Total products:", len(products))
```

## Practice Exercise

A store has monthly sales values for six months:

January: 45000  
February: 52000  
March: 48000  
April: 61000  
May: 57000  
June: 69000  

Tasks:
1. Create a list called months
2. Create a list called monthly_sales
3. Print the first month and its sales value
4. Calculate total sales
5. Calculate average monthly sales
6. Find the highest sales value

```python
# Your Code here
months = ["January", "February", "March", "April", "May", "June"]
monthly_sales = [45000, 52000, 48000, 61000, 57000, 69000]

print("First Month:", months[0])
print("Sales in First Month:", monthly_sales[0])
print("Total Sales:", sum(monthly_sales))
print("Average Monthly Sales:", sum(monthly_sales)/len(monthly_sales))
print("Highest Sales:", max(monthly_sales))
print("Lowest Sales:", min(monthly_sales))
```
