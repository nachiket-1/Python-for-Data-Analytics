## What is Data Analytics?

Data analytics is the process of collecting, organizing, cleaning, analysing and visualizing data, then making decisions based on insights.

analytics_process.png

**Example:** A sales manager may analyze monthly sales data to identify:

- Best-selling products
- Low-performing regions
- Seasonal sales patterns
- Profitability of different products

## Difference between Analysis and Analytics

| Aspect | Analysis | Analytics |
|---|---|---|
| **Meaning** | Examining data to understand what happened | Using data, methods, and tools to generate insights and support decisions |
| **Focus** | Understanding past data | Understanding patterns, predicting outcomes, and improving decisions |
| **Example Question** | Why did sales decrease last month? | Which customers are likely to buy next month? |
| **Approach** | Breaks data into parts and studies it | Uses data analysis, statistics, visualization, and models |
| **Output** | Explanation or summary | Actionable insight, prediction, or recommendation |

## From Data to Decisions

| Product | Units Sold | Selling Price | Cost |
|---|---|---|---|
| Product A | 120 | 500 | 350 |
| Product B | 80 | 700 | 500 |
| Product C | 150 | 300 | 220 |

**Analytics Questions:**

- What is the total revenue?
- Which product generated the highest profit?
- Which product has the best profit margin?
- Which product should the business promote more?

data_to_decisions.png

**Worked answers** (Profit = Units × (Price − Cost); Margin = (Price − Cost) / Price):

| Product | Revenue | Profit | Profit Margin |
|---|---|---|---|
| Product A | 60,000 | 18,000 | 30.0% |
| Product B | 56,000 | 16,000 | 28.6% |
| Product C | 45,000 | 12,000 | 26.7% |

- **Total revenue:** 1,61,000
- **Highest profit and best margin:** Product A, so it is the strongest candidate to promote.

**Decision:** Use the analysis to decide pricing, promotion, and inventory strategy.
