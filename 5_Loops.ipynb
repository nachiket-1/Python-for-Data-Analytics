## Introduction

***Business data usually contains many values.***

Examples:

* Monthly revenue for 12 months
* Ratings from 100 customers
* Performance scores for employees

***Without loops:*** We may need to write repeated code many times.

***With loops:*** Python can repeat the task automatically.

`````python
# Without a loop: repeated code
print("Laptop")
print("Mouse")
print("Keyboard")
`````

With a loop, the same task needs less code:

`````python
products = ["Laptop", "Mouse", "Keyboard"]

for product in products:
    print(product)
`````

## What is a Loop?

* A ***loop*** is used to repeat a block of code multiple times
* If we have a list of products, a loop can be used to go through each product one by one and perform a certain action
* This is useful in analytics, which involves processing multiple records

General form:

````
for each item in a list:
    perform an action
````

## The `for` Loop

* A `for` loop is commonly used to go through values one by one
* The variable `product` takes one value at a time from the list `products`
* For each value, Python runs the indented statement
* ***Indentation*** is important because it tells Python which code belongs inside the loop

Example:

`````python
products = ["Laptop", "Mouse", "Keyboard"]

for product in products:
    print(product)
`````

## Explanation

* In the example above, ***product*** is a ***temporary variable***
* It stores one item from the list at a time
* We can also use other names for the temporary variable (such as ***i***, ***name***, etc.) instead of ***product***
* Thus, the `for` loop takes each item from the `products` list one by one and stores it in the temporary variable `product`

Example:

`````python
products = ["Laptop", "Mouse", "Keyboard"]

for name in products:
    print(name)
`````

## Looping through Sales Values

* Loops can also be used with numeric values
* In this example, Python goes through each sales value in the list and prints it
* The same idea can be extended to calculations such as total sales, performance classification, etc.

Example:

`````python
sales = [50000, 65000, 48000, 72000]

for value in sales:
    print(value)
`````

## Calculating Total Sales

* In this example, we calculate the total sales using a `for` loop
* The `total_sales` variable starts with a value of zero
* We iteratively move through the `sales` list and add the value to `total_sales`

Example:

`````python
sales = [85000, 25000, 32000, 76000]

total_sales = 0

for value in sales:
    total_sales = total_sales + value

print("Total Sales:", total_sales)
`````

**Note**: `total_sales = total_sales + value` can also be written using the assignment operator `+=`, as `total_sales += value`.

## Understanding range()

***range()*** function generates a sequence of numbers.

Example: `range(5)` generates the sequence `0, 1, 2, 3, 4`.

It is commonly used in `for` loops.

Example:

`````python
for i in range(5):
    print(i)
`````

Here, ***i*** is used for index. Its value changes from 0, 1, 2, 3, and 4 as the loop proceeds.

**Note**: `range(5)` stops before 5, so the last number is 4. To see the sequence as a list, use `list(range(5))`.

`````python
print(list(range(5)))
`````

## Understanding range() with len()

* `range()` function is commonly used with `len()` function on lists
* Together, these functions allow us to index and iterate through a list when using a loop

Example:

`````python
my_list = ['First', 'Second', 'Third', 'Fourth']

for i in range(len(my_list)):
    print(my_list[i])
`````

Here:

* `len(my_list)` is 4. Thus, `range(len(my_list))` = `range(4)`
* The value of index `i` varies from 0, 1, 2, 3 as the loop proceeds
* The corresponding element of the list is accessed and printed in each iteration

## Calculating Revenues

* Example: Calculate revenue for multiple products
* In this example, we use two lists: `units_sold` and `selling_prices`
* The revenue is calculated for each item and shown as the output

`````python
units_sold = [100, 50, 80]
selling_prices = [500, 300, 700]

for i in range(len(units_sold)):
    revenue = units_sold[i] * selling_prices[i]
    print(revenue)
`````

## Loop with if-else

Loops can be combined with if-else statements.

Example: For each product, check whether sales are above target

`````python
sales = [85000, 45000, 72000]
target = 60000

for value in sales:
    if value >= target:
        print("Target achieved")
    else:
        print("Target not achieved")
`````

## While Loops (Extra)

**Note**: This section is not in the lecture slides.

A ***while loop*** repeats a block of code ***as long as a condition is true***.

* A `for` loop is used when we know what to go through (a list or a `range`)
* A `while` loop is used when we do not know in advance how many repeats are needed, and want to repeat ***until*** something happens

General form:

````
while condition:
    statement
````

Example: Count from 1 to 5

`````python
count = 1

while count <= 5:
    print(count)
    count += 1
`````

**Note**: The loop variable must change inside the loop (here `count += 1`). Otherwise the condition never becomes `False` and the loop runs forever (an ***infinite loop***).

Business example: How many months until savings reach the target?

`````python
savings = 0
monthly_saving = 15000
goal = 100000
months = 0

while savings < goal:
    savings += monthly_saving
    months += 1

print("Months needed:", months)
print("Savings:", savings)
`````

## More Business Examples (Extra)

Example 1: Count how many products achieved the target

`````python
sales = [85000, 45000, 72000, 30000, 91000]
target = 60000

count = 0

for value in sales:
    if value >= target:
        count += 1

print("Products that achieved target:", count)
`````

Example 2: Collect high sales values into a new list

`````python
sales = [85000, 45000, 72000, 30000, 91000]
target = 60000

high_sales = []

for value in sales:
    if value >= target:
        high_sales.append(value)

print(high_sales)
`````

Example 3: Average customer rating

`````python
ratings = [4.5, 4.2, 3.9, 4.8, 4.0]

total = 0

for rating in ratings:
    total += rating

average = total / len(ratings)
print("Average rating:", round(average, 2))
`````

Example 4: Classify customers using a loop with if-elif-else

`````python
spending = [120000, 65000, 30000, 98000]

for amount in spending:
    if amount >= 100000:
        category = "Premium"
    elif amount >= 50000:
        category = "Regular"
    else:
        category = "Low-value"
    print(amount, "->", category)
`````
