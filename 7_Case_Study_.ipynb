## Customer Churn

* ***Customer churn*** means losing customers over time

Examples:

* Customer stops using a mobile app
* User cancels an OTT subscription
* User shifts to a competitor's service

Retaining existing customers is often easier and cheaper than acquiring new customers.

## What Information Indicates Customer Churn?

A customer may be at risk if:

* Their usage reduced sharply
* They have raised many complaints
* They have given a low satisfaction rating
* Their engagement has dropped

In this case study, we will use three common indicators of churn risk:

***Usage drop, Number of complaints, and Customer rating***

## Input Data

input_data.png

Questions we want to answer:

* Is this customer at risk of customer churn?
* How can we use this data to classify the customer as high, medium or low risk?

## Customer Churn Risk Assessment

churn_risk_rules.png

***A customer is high risk if:***

* Usage drop is at least 30%, ***or***
* Complaints are 3 or more, ***or***
* Rating provided is below 3.0

***A customer is medium risk if:***

* Usage drop is at least 25%, ***or***
* Rating provided is below 4.0

Otherwise, the customer is ***low risk***.

* We will rely on the above business rules to assess churn risk
* In the real world, businesses rely on Machine Learning models considering lots of different customer attributes to accurately predict churn risk

## Recommended Action Based on Risk

recommended_action.png

## Workflow for Case Study

The workflow is as follows:

1. Store customer information
2. Calculate usage drop percentage
3. Classify churn risk
4. Recommend an action

## Step 1: Store Customer Information (Variables)

We store the customer's details in ***variables***.

```python
customer_name = "Asha"
last_month_usage = 40
current_month_usage = 25
complaints = 2
customer_rating = 3.8

print(customer_name, last_month_usage, current_month_usage, complaints, customer_rating)
```

## Step 2: Calculate Usage Drop Percentage (Operators)

Usage drop percentage = (last month usage - current month usage) / last month usage x 100

We use ***arithmetic operators*** for this calculation.

```python
usage_drop = last_month_usage - current_month_usage
usage_drop_percent = usage_drop / last_month_usage * 100

print("Usage drop:", usage_drop, "hours")
print("Usage drop percentage:", usage_drop_percent, "%")
```

## Step 3: Classify Churn Risk (If-Else and Logical Operators)

We use ***comparison operators***, ***logical operators*** (`or`) and an ***if-elif-else*** statement to apply the business rules.

```python
if usage_drop_percent >= 30 or complaints >= 3 or customer_rating < 3.0:
    churn_risk = "High Risk"
elif usage_drop_percent >= 25 or customer_rating < 4.0:
    churn_risk = "Medium Risk"
else:
    churn_risk = "Low Risk"

print("Churn risk:", churn_risk)
```

**Note**: The order of the conditions matters. The `High Risk` rule is checked first. A customer who qualifies as both high and medium risk is therefore classified as high risk.

## Step 4: Recommend an Action

Based on the churn risk, we recommend an action.

```python
if churn_risk == "High Risk":
    action = "Offer retention discount or personal support"
elif churn_risk == "Medium Risk":
    action = "Send engagement message or collect feedback"
else:
    action = "Continue normal engagement"

print("Recommended action:", action)
```

Putting it all together, we can print a summary for the customer:

```python
print("Customer name       :", customer_name)
print("Usage drop          :", round(usage_drop_percent, 1), "%")
print("Complaints          :", complaints)
print("Customer rating     :", customer_rating)
print("Churn risk          :", churn_risk)
print("Recommended action  :", action)
```

## Using Functions

Instead of repeating the code for each customer, we can write ***functions***. A function is written once and reused for any customer.

```python
def calculate_usage_drop_percent(last_month, current_month):
    return (last_month - current_month) / last_month * 100


def classify_churn_risk(usage_drop_percent, complaints, rating):
    if usage_drop_percent >= 30 or complaints >= 3 or rating < 3.0:
        return "High Risk"
    elif usage_drop_percent >= 25 or rating < 4.0:
        return "Medium Risk"
    else:
        return "Low Risk"


def recommend_action(churn_risk):
    if churn_risk == "High Risk":
        return "Offer retention discount or personal support"
    elif churn_risk == "Medium Risk":
        return "Send engagement message or collect feedback"
    else:
        return "Continue normal engagement"
```

Calling the functions for Asha:

```python
drop = calculate_usage_drop_percent(40, 25)
risk = classify_churn_risk(drop, 2, 3.8)
action = recommend_action(risk)

print("Usage drop %:", drop)
print("Churn risk   :", risk)
print("Action       :", action)
```

## Using Lists and Loops

In real data, we have many customers. We store the details in ***lists*** and use a ***loop*** to assess every customer.

```python
names = ["Asha", "Ravi", "Suresh", "Meera", "Karan"]
last_usage = [40, 50, 30, 60, 45]
current_usage = [25, 48, 20, 58, 33]
complaints_list = [2, 0, 4, 1, 0]
ratings = [3.8, 4.5, 4.1, 2.8, 4.7]

for i in range(len(names)):
    drop = calculate_usage_drop_percent(last_usage[i], current_usage[i])
    risk = classify_churn_risk(drop, complaints_list[i], ratings[i])
    action = recommend_action(risk)
    print(names[i], "|", round(drop, 1), "% drop |", risk, "|", action)
```

***Counting customers in each risk category:***

```python
high_count = 0
medium_count = 0
low_count = 0
high_risk_customers = []

for i in range(len(names)):
    drop = calculate_usage_drop_percent(last_usage[i], current_usage[i])
    risk = classify_churn_risk(drop, complaints_list[i], ratings[i])

    if risk == "High Risk":
        high_count += 1
        high_risk_customers.append(names[i])
    elif risk == "Medium Risk":
        medium_count += 1
    else:
        low_count += 1

print("High risk   :", high_count)
print("Medium risk :", medium_count)
print("Low risk    :", low_count)
print("Customers needing immediate attention:", high_risk_customers)
```
