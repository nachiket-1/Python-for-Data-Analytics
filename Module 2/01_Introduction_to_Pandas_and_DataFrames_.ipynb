## Creating a Simple DataFrame

We can create a DataFrame from a ***dictionary***. Each key becomes a column name, and its list of values becomes the column.

Example: Customer service complaint data

```python
import pandas as pd

data = {
    "Complaint ID": ["C101", "C102", "C103"],
    "Customer Type": ["Premium", "Regular", "Premium"],
    "Issue Type": ["Billing", "Delivery", "Technical"],
    "Resolution Time": [4, 12, 7]
}

df = pd.DataFrame(data)
```

## Viewing a DataFrame

We can view the DataFrame by writing its name, or by using `print()`.

Notice the ***index*** (0, 1, 2) on the left. Pandas adds it automatically.

```python
df
```

```python
print(df)
```

## Basic Information about the DataFrame

Useful attributes to understand the structure of the DataFrame:

* `df.shape`: number of rows and columns
* `df.columns`: names of the columns
* `df.index`: the index (row numbers)
* `df.dtypes`: data type of each column

```python
print("Shape  :", df.shape)
print("Columns:", list(df.columns))
print("Index  :", list(df.index))
```

```python
df.dtypes
```

```python
df.info()
```

## Viewing the First Rows

For large datasets, we do not print everything. `df.head()` shows the first 5 rows, and `df.head(2)` shows the first 2 rows.

```python
df["Resolution Time"]
```

To select more than one column, use a ***list of column names***:

```python
df[["Complaint ID", "Resolution Time"]]
```

## Interpreting the Table for Business Decisions

Let us look at the complaint data from a decision-making perspective:

* Which complaint took the longest to resolve?
* What is the average resolution time?
* How many complaints came from premium customers?

```python
print("Longest resolution time :", df["Resolution Time"].max(), "hours")
print("Shortest resolution time:", df["Resolution Time"].min(), "hours")
print("Average resolution time :", round(df["Resolution Time"].mean(), 1), "hours")
```

```python
df["Customer Type"].value_counts()
```

Which complaints took more than 6 hours to resolve? These may need attention.

```python
df[df["Resolution Time"] > 6]
```

**Interpretation**: Complaints C102 (Delivery) and C103 (Technical) took longer than 6 hours. A business can use this to find which issue types are slow to resolve and decide where to improve its service.

## Adding a New Column (Extra)

**Note**: This section is not in the lecture slides.

Just like a formula in Excel, we can create a new column from existing columns. Here we flag whether a complaint was resolved within the target time of 8 hours.

```python
df["Within Target"] = df["Resolution Time"] <= 8
df
```
