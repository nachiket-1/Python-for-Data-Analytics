## What is an Operator?

An ***operator*** is a symbol that performs an operation on values.

Example:

```python
units_sold = 120
selling_price = 500
revenue = units_sold * selling_price

print(revenue)
```

## Here:

* `*` is the ***multiplication operator***
* `units_sold` and `selling_price` are the ***variables***
* `revenue` stores the result

## What is an Expression?

An ***expression*** is a combination of values, variables, and operators that gives a result.

Example:

```python
revenue = 60000
total_cost = 45000

profit = revenue - total_cost
print(profit)
```

## Here:

* `revenue - total_cost` is an ***expression***
* Python evaluates the expression
* The result is stored in `profit`

## Arithmetic Operators

Arithmetic operators are used for mathematical calculations.

(INSERT IMAGE: arithmetic_operators.png here, or use the table below)

| Operator | Meaning | Example |
|:---:|:---:|:---:|
| + | Addition | sales + tax |
| - | Subtraction | revenue - cost |
| * | Multiplication | units * price |
| / | Division | profit / revenue |
| // | Floor division | 105 // 10 = 10 |
| % | Remainder | 105 % 10 = 5 |
| ** | Power | 2 ** 3 = 8 |

```python
sales = 50000
tax = 9000
revenue = 80000
cost = 55000
units = 40
price = 75
profit = 25000

print(sales + tax)       # Addition
print(revenue - cost)    # Subtraction
print(units * price)     # Multiplication
print(profit / revenue)  # Division
print(105 // 10)         # Floor division
print(105 % 10)          # Remainder
print(2 ** 3)            # Power
```

**Note**: `/` always gives a float (decimal) result, while `//` gives only the whole-number part (rounded down).

```python
print(10 / 2)     # 5.0
print(10 // 2)    # 5
print(-7 // 2)    # -4  (rounds down)
```

## Comparison Operators

* Comparison operators compare two values and return `True` or `False`
* Comparison operators are used when we want to check conditions

(INSERT IMAGE: comparison_operators.png here, or use the table below)

| Operator | Meaning | Example |
|:---:|:---:|:---:|
| > | Greater than | sales > target |
| < | Less than | cost < budget |
| >= | Greater than or equal to | rating >= 4.0 |
| <= | Less than or equal to | delay <= 2 |
| == | Equal to | region == "South" |
| != | Not equal to | status != "Inactive" |

**Note**: `=` is used to assign / store a value in a variable. `==` is used to check if two values are equal.

```python
sales = 52000
target = 50000
cost = 40000
budget = 45000
rating = 4.2
delay = 3
region = "South"
status = "Active"

print(sales > target)
print(cost < budget)
print(rating >= 4.0)
print(delay <= 2)
print(region == "South")
print(status != "Inactive")
```

## Logical Operators

Logical operators combine multiple conditions.

(INSERT IMAGE: logical_operators.png here, or use the table below)

| Operator | Meaning | Example |
|:---:|:---:|:---:|
| and | Both conditions must be true | (sales > target) and (rating >= 4) |
| or | At least one condition must be true | (sales > target) or (rating >= 4) |
| not | Reverses the condition | not customer_active |

***Boolean vs Logical Operators:***

Logical operators are commonly used on Boolean values (`True` / `False`).

For example: if condition1 `and` condition2 are `True`, perform some action.

```python
sales = 52000
target = 50000
rating = 3.8
customer_active = True

print((sales > target) and (rating >= 4))
print((sales > target) or (rating >= 4))
print(not customer_active)
```

## Assignment Operators

Assignment operators are used to assign or update values.

(INSERT IMAGE: assignment_operators.png here, or use the table below)

| Operator | Meaning | Example | Interpretation |
|:---:|:---:|:---:|:---|
| = | Assign value | sales = 50000 | 50000 is assigned to sales |
| += | Add and update | sales += 10000 | 10000 is added to the current value of sales. Same as sales = sales + 10000, so sales = 60000 |
| -= | Subtract and update | sales -= 10000 | 10000 is subtracted from the current value of sales. Same as sales = sales - 10000, so sales = 50000 |
| *= | Multiply and update | price *= 1.10 | price is multiplied by 1.1 and the new value is stored in price again |
| /= | Divide and update | budget /= 2 | budget is divided by 2 and the new value is stored in budget again |

```python
sales = 50000
sales += 10000
print(sales)      # 60000

sales -= 10000
print(sales)      # 50000

price = 100
price *= 1.10
print(price)      # about 110

budget = 80000
budget /= 2
print(budget)     # 40000.0
```

## Order of Operations

Python follows the mathematical order of operations.

***Order:***

1. Parentheses `()`
2. Power `**`
3. Multiplication, division, remainder
4. Addition and subtraction

Example 1:

```python
result = 10 + 5 * 2
print(result)
```

Example 2:**bold text**

```python
result = (10 + 5) * 2
print(result)
```

## Examples

Example 1: Sales report

```python
units_sold = 150
selling_price = 200
total_cost = 22000

revenue = units_sold * selling_price
profit = revenue - total_cost
profit_margin = profit / revenue

print("Revenue:", revenue)
print("Profit:", profit)
print("Profit margin:", round(profit_margin * 100, 2), "%")
```

Example 2: Bonus eligibility (comparison + logical)

```python
sales = 55000
target = 50000
rating = 4.5

eligible = (sales > target) and (rating >= 4)
print("Eligible for bonus:", eligible)
```

Example 3: Floor division and remainder

```python
total_minutes = 275
hours = total_minutes // 60
minutes = total_minutes % 60

print(hours, "hours", minutes, "minutes")
```
