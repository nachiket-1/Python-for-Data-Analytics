# What is a variable?

A ***variable*** is a name used to store a value.

Example:

```python
sales = 50000
```

###Here:

***Sales*** is a variable name

50000 is the valuestored in the variable

##Why do we need variable?

***Variables helps us:***

*   Store business values
*   Reuse values for calculations
*   Make code easier to understand
*   Avoid repeating numbers again and again
*   Update values easily







**Without variables:**

```python
print(120*50)
```

**With varibales:**

```python
units_sold = 120
selling_price = 500
revenue = units_sold * selling_price

print(revenue)
```

##Variable naming rules

***Rules for naming varibales in python:***

*   Variable names can contain letter,      numbers, and underscore
*   They cannot start with a number
*   They cannot contain spaces
*   They are case-sensitive
*   Use meaningfull names






**Good variable names:**

*   sales
*   total_revenue
*   customer_rating
*   profit_margin


**Avoid:**


*   x
*   total revenue
*   2 sales



##Common Data types in Pyhton

***Data Type*** means the kind of value tored in a variable

1. ***Integer***

Integer stores whole numbers.

Example:

*   units_sold = 250
*   income = 50000
*   employee_count = 420

In business analytics, integer values are very common because we often count transactions, orders, products, and customers.







2. ***Float***

Float (float) stores decimal values.

Example:

*   profit_margin = 19.2
*   average_rating = 4.9
*   discount_rate = 15.3

In busniness analytics, decimal values are important because averages, percentages, ratios, and rates usually usally contain decimal values.






3. ***String***

String (str) stores text values.

Example:

*   product_name = "laptop"
*   region = "south"
*   customer_segment = 'premium'

**Note**: Text values should be inside single (' ') or double (" ") quatation marks.

Dataset cointains many text fields such as product names, customers names, city names, branch names,  or customer segmentation.





4. ***Boolean***

Boolean (bool) stores only two values:

1.   True
2.   False

Example:

*   target_achieved = True
*   payments_received = True
*   customer_active = False

**Business use**:

*   Has employee achieved the target?
*   Has customer paid?
*   Is the product available?
*   Is the customer active?
